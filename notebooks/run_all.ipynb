```python
STEPS = ['01_checkpoints','02_inference','03_ood','04_corruption',
         '05_explanations','06_adversarial','07_probes','08_analysis']
status = {}
for s in STEPS:
    try:
        info = nb.run(s)
        status[s] = 'ok' if info.get('ok') else 'failed'
    except Exception as e:
        status[s] = f'error: {e}'
        print('ABORTING after', s, ':', e)
        break
print(json.dumps(status, indent=2))
!cd /content && zip -qr /content/drive/MyDrive/arch-complementarity.zip out
print('ZIP written')
```
