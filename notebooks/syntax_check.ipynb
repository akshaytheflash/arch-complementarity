# Syntax check — CPU only, ~30 s

Compiles every module in `src/` so a syntax error is caught before the long GPU steps.

```python
import py_compile, glob, os, sys, traceback
!git clone -q https://github.com/akshaytheflash/arch-complementarity.git /content/repo 2>/dev/null || (cd /content/repo && git pull -q)
bad = []
for f in sorted(glob.glob('/content/repo/src/**/*.py', recursive=True)):
    try:
        py_compile.compile(f, doraise=True)
        print('OK  ', os.path.relpath(f, '/content/repo'))
    except Exception as e:
        bad.append((f, str(e)))
        print('FAIL', os.path.relpath(f, '/content/repo'), e)
print('BAD_COUNT', len(bad))
for f, e in bad:
    print('---', f); print(e[:600])
print('SYNTAX_CHECK_DONE')
```
